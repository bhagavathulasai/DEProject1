```python
import numpy as np
import cv2
import os
from sklearn.model_selection import StratifiedShuffleSplit
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from tensorflow.keras.applications import ResNet50, MobileNet, EfficientNetB0
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt

# Paths and Parameters
path = "/content/drive/MyDrive/67/3"  # Update dataset path
save_path = "/content/drive/MyDrive/ModelStorage"  # Path to save models
image_dimensions = (128, 128)  # Image size (height, width)
test_ratio = 0.2
validation_ratio = 0.2

# Ensure dataset path exists
if not os.path.exists(path):
    raise FileNotFoundError(f"Dataset path '{path}' not found.")

# Create save directory if it does not exist
os.makedirs(save_path, exist_ok=True)

# Data Loading
images, labels, classNames = [], [], []

def load_images_for_class(classDir, classIndex):
    class_images, class_labels = [], []
    for imgName in os.listdir(classDir):
        if imgName.lower().endswith((".png", ".jpg", ".jpeg")):
            img_path = os.path.join(classDir, imgName)
            curImg = cv2.imread(img_path)
            if curImg is not None:
                curImg = cv2.resize(curImg, image_dimensions)
                class_images.append(curImg)
                class_labels.append(classIndex)
    return class_images, class_labels

print("Loading dataset...")
myList = sorted([item for item in os.listdir(path) if os.path.isdir(os.path.join(path, item))])
noOfClasses = len(myList)

if noOfClasses == 0:
    raise ValueError("No valid classes found in the dataset path.")

print(f"Total Classes Detected: {noOfClasses}")

for count, classDir in enumerate(myList):
    classDirPath = os.path.join(path, classDir)
    class_images, class_labels = load_images_for_class(classDirPath, count)
    if class_images:
        classNames.append(myList[count])
        images.extend(class_images)
        labels.extend(class_labels)

if not images:
    raise ValueError("No images were loaded.")

print("Data import complete.")
images = np.array(images)
labels = np.array(labels)

# Normalize images
images = images / 255.0

# Check image shape
print(f"Images shape: {images.shape}")

# Splitting Data
sss = StratifiedShuffleSplit(n_splits=1, test_size=test_ratio, random_state=42)
for train_index, test_index in sss.split(images, labels):
    X_train, X_test = images[train_index], images[test_index]
    y_train, y_test = labels[train_index], labels[test_index]

# Validation Split
sss_validation = StratifiedShuffleSplit(n_splits=1, test_size=validation_ratio, random_state=42)
for train_index, validation_index in sss_validation.split(X_train, y_train):
    X_train_final, X_validation = X_train[train_index], X_train[validation_index]
    y_train_final, y_validation = y_train[train_index], y_train[validation_index]

# Data Augmentation
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

datagen.fit(X_train_final)

# Improved CNN Architecture
def build_cnn_improved():
    model = Sequential()
    model.add(Conv2D(64, (3, 3), activation='relu', input_shape=(image_dimensions[0], image_dimensions[1], 3)))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(BatchNormalization())

    model.add(Conv2D(128, (3, 3), activation='relu'))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(BatchNormalization())

    model.add(Conv2D(256, (3, 3), activation='relu'))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(BatchNormalization())

    model.add(Flatten())
    model.add(Dense(512, activation='relu'))
    model.add(Dropout(0.5))
    model.add(Dense(noOfClasses, activation='softmax'))
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Transfer Learning Models
def build_resnet50():
    base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(128, 128, 3))
    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.5)(x)
    predictions = Dense(noOfClasses, activation='softmax')(x)

    model = Model(inputs=base_model.input, outputs=predictions)

    # Freeze base model layers
    for layer in base_model.layers:
        layer.trainable = False

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def build_mobilenet():
    base_model = MobileNet(weights='imagenet', include_top=False, input_shape=(128, 128, 3))
    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.5)(x)
    predictions = Dense(noOfClasses, activation='softmax')(x)

    model = Model(inputs=base_model.input, outputs=predictions)

    # Freeze base model layers
    for layer in base_model.layers:
        layer.trainable = False

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

def build_efficientnet():
    base_model = EfficientNetB0(weights='imagenet', include_top=False, input_shape=(128, 128, 3))
    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.5)(x)
    predictions = Dense(noOfClasses, activation='softmax')(x)

    model = Model(inputs=base_model.input, outputs=predictions)

    # Freeze base model layers
    for layer in base_model.layers:
        layer.trainable = False

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Train and Evaluate Models
models = {
    "Improved_CNN": build_cnn_improved(),
    "ResNet50": build_resnet50(),
    "MobileNet": build_mobilenet(),
    "EfficientNet": build_efficientnet()
}

histories = {}

for model_name, model in models.items():
    print(f"\nTraining {model_name}...")
    history = model.fit(
        datagen.flow(X_train_final, y_train_final, batch_size=32),
        validation_data=(X_validation, y_validation),
        epochs=20,
        callbacks=[EarlyStopping(monitor='val_loss', patience=3)]
    )

    # Evaluate
    test_loss, test_acc = model.evaluate(X_test, y_test)
    print(f"{model_name} Test Accuracy: {test_acc * 100:.2f}%")

    # Save Model
    model.save(os.path.join(save_path, f"{model_name}_Model.h5"))
    print(f"{model_name} model saved.")

    histories[model_name] = history

# Plot Accuracy Graphs
plt.figure(figsize=(12, 8))
for model_name, history in histories.items():
    plt.plot(history.history['accuracy'], label=f'{model_name} Training Accuracy')


plt.title('Model Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
